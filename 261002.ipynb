```python
import random

a = random.randint(10, 99)
b = random.randint(2, 9)

print(f'{a}를 {b}로 나누면 몫과 나머지는 얼마일까요?')

my_quotient = int(input('몫: '))
my_remainder = int(input('나머지: '))

quotient = a // b
remainder = a% b

print(f'정답 - 몫: {quotient}, 나머지: {remainder}')
print('몫 정답 여부:', my_quotient == quotient)
print('나머지 정답 여부:', my_remainder == remainder)
```

```python
import random

dice1 = random.randint(1, 6)
dice2 = random.randint(1, 6)

print(f'첫 번째 주사위: {dice1}')
print(f'두 번째 주사위: {dice2}')

if dice1 > dice2:
    print('첫 번째 주사위가 큽니다.')
elif dice1 < dice2:
    print('두 번째 주사위가 큽니다.')
else:
    print('두 주사위가 같습니다.')
```

```python
import random

# 1: 가위, 2: 바위, 3: 보
com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == com:
    print('비겼습니다.')
elif user == 1 and com == 3:
    print('이겼습니다!')   # 가위 vs 보
elif user == 2 and com == 1:
    print('이겼습니다!')   # 바위 vs 가위
elif user == 3 and com == 2:
    print('이겼습니다!')   # 보 vs 바위
else:
    print('졌습니다.')
```

```python
def add(x, y):
    return x + y

def sub(x, y):
    return x - y

def mul(x, y):
    return x * y

def div(x, y):
    return x / y

def asmd(x,y,op):
    if op == '+':
        print(f'{x} + {y} = {add(x, y)}')
    elif op == '-':
        print(f'{x} - {y} = {sub(x, y)}')
    elif op == '*':
        print(f'{x} * {y} = {mul(x, y)}')
    elif op == '/':
        print(f'{x} / {y} = {div(x, y):.2f}')
    else:
        print("잘못된 연산자입니다.")     
```

```python
x = int(input('첫 번째 수: '))
y = int(input('두 번째 수: '))
op = input('연산자를 입력하시오')
asmd(x,y,op)
```
