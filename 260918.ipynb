```python
#숫자를 입력받아서 짝수인지 홀수인지 출력하는 프로그램을 작성하시오
num = int(input('숫자를 입력하세요:'))
number = num%2

if number > 0:
    print(num, '은/는 홀수 입니다')
else:
    print(num, '은/는 짝수 입니다')
```
